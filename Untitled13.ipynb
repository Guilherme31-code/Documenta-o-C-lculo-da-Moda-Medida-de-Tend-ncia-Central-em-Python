Os Métodos da Classe Dicionário

A classe dict no Python possui um conjunto de métodos embutidos muito úteis que facilitam a manipulação dos dados. Para facilitar o entendimento, podemos dividi-los em quatro categorias principais: Acesso, Atualização, Remoção e Utilitários.

1. Métodos de Acesso e Visualização

Estes métodos são usados para buscar informações dentro do dicionário sem alterá-lo.

get(chave, [valor_padrao]): Retorna o valor de uma chave específica. Se a chave não existir, retorna None (ou um valor padrão que você definir), evitando que seu código quebre com um erro KeyError.



```python
carro = {"marca": "Ford", "modelo": "Mustang"}
print(carro.get("ano", "Ano não encontrado")) # Saída: Ano não encontrado
```

keys(): Retorna uma visualização (como uma lista) contendo todas as chaves do dicionário.

values(): Retorna uma visualização contendo todos os valores do dicionário.

items(): Retorna uma visualização contendo pares de (chave, valor) em formato de tuplas. É excelente para usar em loops for.

2. Métodos de Adição e Atualização

sados para inserir novos dados ou modificar os existentes de forma mais avançada.

```python
carro = {"marca": "Ford", "modelo": "Mustang"}
carro.update({"ano": 1964})
print(carro) # Saída: {'marca': 'Ford', 'modelo': 'Mustang', 'ano': 1964}
```

setdefault(chave, [valor_padrao]): Retorna o valor de uma chave se ela existir. Se não existir, ele cria a chave no dicionário atribuindo a ela o valor_padrao (ou None se nada for passado) e retorna esse valor.

```python
carro = {"marca": "ford"}
carro.sedefault("cor", "preto")
```

3. Métodos de Remoção

Servem para deletar itens ou esvaziar o dicionário completamente.

popitem(): Remove e retorna o último par (chave, valor) inserido no dicionário. (Em versões do Python anteriores à 3.7, ele removia um item aleatório).

clear(): Esvazia o dicionário por completo, removendo todos os seus iten

```python
carro.clear()
# carro agora é: {}
```

4. Métodos Utilitários (Cópia e Criação)

copy(): Retorna uma cópia rasa (shallow copy) do dicionário. Isso é importante porque se você fizer apenas dict2 = dict1, ambos apontarão para o mesmo lugar na memória, e alterar um mudará o outro. Usar .copy() evita isso.

```python
carro = {"marca": "Ford"}
carro_copia =  carro.copy()
```

fromkeys(iteravel, [valor]): Cria e retorna um novo dicionário usando as chaves de um iterável (como uma lista ou tupla) e atribui a todas elas o mesmo valor (que por padrão é None).

```python
chaves = ("notas1", "'notass2", "notas3")
notass = dict.fromkeys(chaves, 0)
```
