Analisando a afirmação: "Em Python tudo é objeto"


Em Python, a afirmação "tudo é objeto" significa que qualquer entidade no programa, números, strings, listas, funções, classes, módulos, etc. é uma instância de alguma classe. Ou seja, cada valor ou elemento do código possui:

Identidade (um endereço na memória, acessível via id())

Tipo (definido pela classe, acessível via type())

Atributos e métodos (comportamentos associados)

Exemplos práticos

Tipos básicos são objetos

```python
x = 42          # int é uma classe
print(type(x))  # <class 'int'>
print(x.bit_length())  # método do objeto int
```

Funções são objetos

```python
def ola():
    return "Olá"

print(type(ola))   # <class 'function'>
ola.nova_attr = 1  # podemos adicionar atributos dinamicamente
```

Classes também são objetos

```python
class MinhaClasse:
    pass

print(type(MinhaClasse))  # <class 'type'> (metaclasse)
MinhaClasse.valor = 100   # atributo de classe
```

Módulos são objetos

```python
import math
print(type(math))         # <class 'module'>
print(math.pi)            # acesso a atributo
```

Consequências importantes

Atribuição por referência
Quando você faz b = a, ambas as variáveis apontam para o mesmo objeto. Alterações no objeto refletem em todas as referências

Tipagem dinâmica
Como tudo é objeto, uma variável pode referenciar qualquer tipo de objeto a qualquer momento.

Polimorfismo facilitado
Qualquer objeto pode ser passado como argumento para uma função, desde que implemente os métodos necessários

Possibilidade de introspecção
Você pode inspecionar e modificar dinamicamente atributos de objetos em tempo de execução.

Exceção aparente

Alguns tipos imutáveis (como int, str, tuple) são objetos, mas seus valores não podem ser alterados após a criação. Ainda assim, eles obedecem plenamente à definição de objeto.

Portanto, a frase resume um dos pilares da filosofia Python: uniformidade na manipulação de dados e código, permitindo flexibilidade, reflexão e metaprogramação
