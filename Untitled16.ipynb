SISTEMA DE REGISTRO DE BICICLETA


# 🚲 Sistema de Registro de Vendas de Bicicletas

## Descrição

João possui uma bicicletaria e deseja um sistema para registrar as informações de suas bicicletas.

Cada bicicleta deve possuir:

- Modelo
- Cor
- Ano
- Valor

Além disso, uma bicicleta deve ser capaz de:

- Buzinar
- Parar
- Correr

O projeto foi desenvolvido utilizando Programação Orientada a Objetos (POO) em Python.

---

## Estrutura da Classe

### Atributos

| Atributo | Descrição |
|-----------|------------|
| modelo | Modelo da bicicleta |
| cor | Cor da bicicleta |
| ano | Ano de fabricação |
| valor | Valor da bicicleta |

### Métodos

| Método | Função |
|----------|---------|
| buzinar() | Emite o som da buzina |
| correr() | Simula a bicicleta em movimento |
| parar() | Simula a parada da bicicleta |

---

## Exemplo de Saída

Modelo: Caloi Elite  
Cor: Preta  
Ano: 2024  
Valor: R$ 3500.00

Bi Bi!  
A bicicleta está correndo.  
A bicicleta parou.

---

## Conceitos Aplicados

- Classe
- Objeto
- Atributos
- Métodos
- Instanciação
- Encapsulamento

---



Projeto desenvolvido para estudo de Programação Orientada a Objetos em PythoN

```python
class Bicicleta:
    def __init__(self, modelo, cor, ano, valor):
        self.modelo = modelo
        self.cor = cor
        self.ano = ano
        self.valor = valor

    def buzinar(self):
        print("Bi Bi!")

    def correr(self):
        print("A bicicleta está correndo.")

    def parar(self):
        print("A bicicleta parou.")


bicicleta1 = Bicicleta(
    "Caloi Elite",
    "Preta",
    2024,
    3500.00
)

print(f"Modelo: {bicicleta1.modelo}")
print(f"Cor: {bicicleta1.cor}")
print(f"Ano: {bicicleta1.ano}")
print(f"Valor: R$ {bicicleta1.valor:.2f}")

bicicleta1.buzinar()
bicicleta1.correr()
bicicleta1.parar()
```

EXPLICAÇÃO DO CÓDIGO

1. A classe Bicicleta é criada para representar uma bicicleta.

2. O método __init__() é o construtor da classe.
   Ele recebe os atributos:
   - modelo
   - cor
   - ano
   - valor

3. Os métodos buzinar(), correr() e parar()
   representam os comportamentos da bicicleta.

4. bicicleta1 é um objeto criado a partir da classe Bicicleta.

5. Os atributos são exibidos utilizando f-strings.

6. Os métodos são chamados para simular as ações da bicicleta.
