Sistema de Biblioteca


#Função para carregar livros


```python
def carregar_livros():
  try:
    with open("livros.txt", "r") as arquivo:
      livros = {}
      for linha in arquivo:
        título, autor = linha.strip().split(",")
        livros[título] = autor
      return livros
  except FileNotFoundError:
    return {}
```

#Função para cadastrar livros


```python
def cadastrar_livro(livros):
  título = input("Digite o título do livro: ")
  if título in livros:
    print("Livro já cadastrado.")
  else:
    autor = input("Digite o autor do livro: ")
    livros[título] = autor
    with open("livros.txt", "a") as arquivo:
      arquivo.write(f"{título},{autor}\n")
    print("Livro cadastrado com sucesso.")
  return livros
```

#Função para listar livros

```python
def listar_livros(livros):
  if not livros:
    print("Nenhum livro cadastrado.")
  else:
    print("Livros cadastrados:")
    for título, autor in livros.items():
      print(f"{título} - {autor}")
```

#Função para remover livros


```python
def remover_livros(livros):
  titulo = input("Digite o título do livro que deseja remover: ")
  if titulo in livros:
    del livros[titulo]
    with open("livros.txt", "w") as arquivo:
      for t, a in livros.items():
        arquivo.write(f"{t},{a}\n")
    print("Livro removido com sucesso.")
  else:
    print("Livro não encontrado")
  return livros
```

#Função para alterar o autor de um livro

```python
def alterar_livro(livros):
  titulo = input("Digite o título do livro que deseja alterar: ")
  if titulo in livros:
    novo_autor = input("Digite o novo autor do livro: ")
    livros[titulo] = novo_autor
    with open("livros.txt", "w") as arquivo:
      for t, a in livros.items():
        arquivo.write(f"{t},{a}\n")
    print("Autor alterado com sucesso.")
  else:
    print("Livro não encontrado.")
  return livros
```

#Menu principal

```python
def menu_principal():
  livros = carregar_livros() # Call the function
  while True: # Corrected boolean
    print("\nMenu Principal:")
    print("1. Cadastrar Livro")
    print("2. Listar Livros")
    print("3. Remover Livro")
    print("4. Alterar Autor de Livro")
    print("5. Sair")
    escolha = input("Escolha uma opção: ") # Use 'escolha' consistently
    if escolha == "1":
      livros = cadastrar_livro(livros)
    elif escolha == "2":
      listar_livros(livros)
    elif escolha == "3":
      livros = remover_livros(livros)
    elif escolha == "4":
      livros = alterar_livro(livros)
    elif escolha == "5":
      print("Saindo...")
      break
    else:
      print("Opção inválida.")
```

#Executa o programa

```python
menu_principal()
```
